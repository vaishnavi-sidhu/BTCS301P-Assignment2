```python
from google.colab import files
uploaded = files.upload()   # yaha apni image select karo
image_path = list(uploaded.keys())[0]
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread(image_path)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(5,5))
plt.imshow(gray, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.hist(gray.ravel(), bins=256, range=[0,256], color='gray')
plt.title("Histogram - Original Image")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def contrast_stretch(img):
    min_val = np.min(img)
    max_val = np.max(img)
    stretched = (img - min_val) * (255.0 / (max_val - min_val))
    return stretched.astype(np.uint8)

stretched_img = contrast_stretch(gray)

plt.figure(figsize=(5,5))
plt.imshow(stretched_img, cmap='gray')
plt.title("Contrast Stretched Image")
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

equalized_img = cv2.equalizeHist(gray)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(gray, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(equalized_img, cmap='gray')
plt.title("Histogram Equalized")
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
clahe_img = clahe.apply(gray)

plt.figure(figsize=(5,5))
plt.imshow(clahe_img, cmap='gray')
plt.title("CLAHE Enhanced Image")
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

fig, axs = plt.subplots(1, 3, figsize=(15,4))

axs[0].hist(gray.ravel(), bins=256, range=[0,256], color='gray')
axs[0].set_title("Original Histogram")

axs[1].hist(equalized_img.ravel(), bins=256, range=[0,256], color='blue')
axs[1].set_title("Equalized Histogram")

axs[2].hist(clahe_img.ravel(), bins=256, range=[0,256], color='green')
axs[2].set_title("CLAHE Histogram")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

fig, axs = plt.subplots(1, 4, figsize=(18,5))

axs[0].imshow(gray, cmap='gray'); axs[0].set_title("Original"); axs[0].axis('off')
axs[1].imshow(stretched_img, cmap='gray'); axs[1].set_title("Contrast Stretched"); axs[1].axis('off')
axs[2].imshow(equalized_img, cmap='gray'); axs[2].set_title("Histogram Equalized"); axs[2].axis('off')
axs[3].imshow(clahe_img, cmap='gray'); axs[3].set_title("CLAHE"); axs[3].axis('off')

plt.tight_layout()
plt.show()
```
